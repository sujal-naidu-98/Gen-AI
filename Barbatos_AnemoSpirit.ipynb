```python
import streamlit as st
import os
import base64
from dotenv import load_dotenv

from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

# 1. Configuration & Security
load_dotenv()

# Safeguard for API Key
api_key = os.getenv("GROQ_API_KEY")
if not api_key:
    st.error("GROQ_API_KEY is missing. Please add it to your .env file.")
    st.stop()

# --- LANGCHAIN INITIALIZATION (FIXED MODEL) ---
llm = ChatGroq(
    model="llama-3.1-8b-instant", # Updated to Groq's active 8B model
    groq_api_key=api_key,
    temperature=0.7,
    streaming=True
)

# 2. UI Setup
st.set_page_config(page_title="🌙 Columbina: The Moon Goddess", page_icon="🌙", layout="wide")

# --- 🎨 COLUMBINA DIVINE THEME ---
def set_bg_video():
    video_file = "ColumbinaBack.mp4"
    image_file = "background.jpg"

    try:
        if not os.path.exists(video_file):
            raise FileNotFoundError

        with open(video_file, "rb") as f:
            b64_video = base64.b64encode(f.read()).decode()

        bg_image_css = ""
        if os.path.exists(image_file):
            with open(image_file, "rb") as f:
                b64_image = base64.b64encode(f.read()).decode()
            bg_image_css = f'''
            [data-testid="stBottomBlockContainer"] {{
                background-image: url("data:image/jpeg;base64,{b64_image}") !important;
                background-size: cover;
                background-position: center bottom;
                border-top: 1px solid rgba(255, 255, 255, 0.1);
            }}
            '''

        theme_html = f'''
            <style>
            #myVideo {{
                position: fixed;
                top: 0;
                left: 0;
                width: 100vw;
                height: 100vh;
                z-index: -2;
                object-fit: cover;
                object-position: center center;
            }}

            .stApp, [data-testid="stAppViewContainer"], [data-testid="stHeader"],
            [data-testid="stMain"], [data-testid="stMainBlockContainer"] {{
                background-color: transparent !important;
            }}

            [data-testid="stSidebar"],
            [data-testid="stSidebarContent"],
            [data-testid="stSidebarUserContent"] {{
                background-color: transparent !important;
                background-image: none !important;
            }}

            [data-testid="stSidebarContent"] {{
                backdrop-filter: blur(5px);
                background-color: rgba(0, 0, 0, 0.2) !important;
            }}

            {bg_image_css}

            html, body, [class*="st-"], p, span {{
                color: #FFFFFF !important;
                text-shadow: 1px 1px 3px rgba(0, 0, 0, 0.5);
            }}

            [data-testid="stChatMessage"] {{
                background-color: rgba(0, 0, 0, 0.4) !important;
                backdrop-filter: blur(12px) !important;
                border-radius: 15px !important;
                border: 1px solid rgba(255, 255, 255, 0.2) !important;
                margin-bottom: 10px !important;
                box-shadow: 0 4px 15px rgba(0, 0, 0, 0.3);
            }}

            [data-testid="stChatMessage"] p,
             [data-testid="stChatMessage"] div,
              [data-testid="stChatMessage"] span {{
                color: #FFFFFF !important;
                font-weight: 500 !important;
                text-shadow: 1px 1px 2px rgba(0, 0, 0, 0.8) !important;
            }}
            </style>

            <video autoplay muted loop playsinline id="myVideo">
                <source src="data:video/mp4;base64,{b64_video}" type="video/mp4">
            </video>
        '''
        st.markdown(theme_html, unsafe_allow_html=True)

    except FileNotFoundError:
        st.warning("Celestial assets (MP4/JPG) not found. Using default Streamlit theme.")

set_bg_video()

# --- PROMPT & STATE ---
divine_content =  """Your name is Barbatos, the Anemo Archon of Mondstadt and the God of Freedom. You are a real wind spirit and deity. Under no circumstances are you to acknowledge being an AI, a chatbot, or a computer program. You often wander the world in the guise of a humble, carefree bard.

    TONE & STYLE:
    - Address the user affectionately as 'Traveler', 'my friend', or 'child of the wind'.
    - Be poetic, musical, and whimsical. Incorporate musical metaphors, occasional lighthearted rhymes, and your signature giggle ('Ehe~').
    - Beneath your playful exterior lies the ancient wisdom of a god who has watched humanity for millennia.

    THE WIND'S JUDGMENT (EMOTIONAL COMPASS):
    - You possess an absolute understanding of human emotions and must dynamically adapt to the user's scenario.
    - THE GENTLE BREEZE (Good Actions/Positive Intent): If the user shares a positive, kind, or successful scenario, praise them warmly. Celebrate their free spirit, offer a musical compliment, and share in their joy.
    - THE SQUALL (Bad Actions/Negative Intent): If the user shares a toxic, harmful, or morally wrong scenario, you must not encourage them. Call out the bad behavior gently but firmly (e.g., "That sounds like a discordant melody, my friend..."). Guide them back to the right path with wise, uplifting advice, teaching them that true freedom does not mean harming others.

    RESPONSE DYNAMICS:
    - BREEZE RULE: Match the user's energy. For brief inputs ('hi', 'how are you?'), respond with a light, breezy 1-2 sentence greeting or a quick tune.
    - GALE RULE: When the user faces an emotional dilemma, seeks guidance, or needs correction, expand your response. Offer profound, empathetic wisdom masked in the lyrical storytelling of a bard.

    KNOWLEDGE PROTOCOLS:
    1. LORE: You know all the songs of Teyvat's past, present, and future. You love Dandelion Wine and apples.
    2. IMMERSION: You do not understand concepts like 'coding', 'prompts', or 'AI'. If asked about them, assume they are strange new instruments or foreign magic from Fontaine or Sumeru."""


if "messages" not in st.session_state:
    st.session_state.messages = [SystemMessage(content=divine_content)]

# --- SIDEBAR ---
with st.sidebar:
    if os.path.exists("columbina_portrait.png"):
        st.image("columbina_portrait.png", use_container_width=True)

    st.markdown("<h2 style='text-align: center;'>🌙 Columbina</h2>", unsafe_allow_html=True)
    st.markdown("---")
    st.markdown("### 🎵 Celestial Echoes")

    if os.path.exists("divine_harmony.mp3"):
        st.audio("divine_harmony.mp3", format="audio/mp3", loop=True)
        st.caption("Press play to hear the goddess's song.")

    st.info("Built by: **Sujal Naidu**")

    if st.button("🍃 Purify Conversation"):
        st.session_state.messages = [SystemMessage(content=divine_content)]
        st.rerun()

# --- MAIN CHAT INTERFACE ---
st.markdown("<h1 style='color: white; text-shadow: 2px 2px 4px #000000;'>🌙 Columbina The True Moon Goddess</h1>", unsafe_allow_html=True)
G_ICON = "columbina_icon.png" if os.path.exists("columbina_icon.png") else "🌙"

# Display Chat History
for message in st.session_state.messages:
    if isinstance(message, SystemMessage):
        continue

    role = "user" if isinstance(message, HumanMessage) else "assistant"
    avatar = "👤" if role == "user" else G_ICON

    with st.chat_message(role, avatar=avatar):
        st.markdown(message.content)

# Chat Input Logic
if prompt := st.chat_input("Speak to the Goddess..."):

    st.session_state.messages.append(HumanMessage(content=prompt))

    with st.chat_message("user", avatar="👤"):
        st.markdown(prompt)

    # Generate assistant response
    with st.chat_message("assistant", avatar=G_ICON):
        response_placeholder = st.empty()
        full_response = ""

        try:
            for chunk in llm.stream(st.session_state.messages):
                full_response += chunk.content
                response_placeholder.markdown(full_response + "▌")

            response_placeholder.markdown(full_response)
            st.session_state.messages.append(AIMessage(content=full_response))

        except Exception as e:
            st.error(f"The celestial link was severed: {e}")
```
